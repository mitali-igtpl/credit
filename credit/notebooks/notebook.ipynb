```python
previous_df['NAME_TYPE_SUITE'] = (
    previous_df['NAME_TYPE_SUITE'].fillna('Unknown')
)
```

```python
prev_fe = previous_df.copy()
```

```python
prev_features = (
    prev_fe.groupby('SK_ID_CURR')
    .agg(
        PREV_APP_COUNT=('SK_ID_PREV', 'count')
    )
    .reset_index()
)
```

```python
numeric_features = (
    prev_fe.groupby('SK_ID_CURR')
    .agg(
        PREV_AMT_APPLICATION_MEAN=('AMT_APPLICATION', 'mean'),
        PREV_AMT_APPLICATION_MAX=('AMT_APPLICATION', 'max'),

        PREV_AMT_CREDIT_MEAN=('AMT_CREDIT', 'mean'),
        PREV_AMT_CREDIT_MAX=('AMT_CREDIT', 'max'),

        PREV_AMT_ANNUITY_MEAN=('AMT_ANNUITY', 'mean'),
        PREV_AMT_GOODS_PRICE_MEAN=('AMT_GOODS_PRICE', 'mean'),

        PREV_CNT_PAYMENT_MEAN=('CNT_PAYMENT', 'mean'),

        PREV_AMT_DOWN_PAYMENT_MEAN=('AMT_DOWN_PAYMENT', 'mean'),
        PREV_RATE_DOWN_PAYMENT_MEAN=('RATE_DOWN_PAYMENT', 'mean')
    )
    .reset_index()
)
```

```python
prev_features = prev_features.merge(
    numeric_features,
    on='SK_ID_CURR',
    how='left'
)
```

```python
status_features = (
    pd.crosstab(
        prev_fe['SK_ID_CURR'],
        prev_fe['NAME_CONTRACT_STATUS']
    )
    .reset_index()
)
```

```python
status_features.columns = [
    'SK_ID_CURR'
] + [
    f'PREV_STATUS_{col.upper().replace(" ", "_")}'
    for col in status_features.columns[1:]
]
```

```python
prev_features = prev_features.merge(
    status_features,
    on='SK_ID_CURR',
    how='left'
)
```

```python
if 'PREV_STATUS_APPROVED' in prev_features.columns:
    prev_features['PREV_APPROVAL_RATE'] = (
        prev_features['PREV_STATUS_APPROVED']
        / prev_features['PREV_APP_COUNT']
    )
```

```python
if 'PREV_STATUS_REFUSED' in prev_features.columns:
    prev_features['PREV_REFUSAL_RATE'] = (
        prev_features['PREV_STATUS_REFUSED']
        / prev_features['PREV_APP_COUNT']
    )
```

```python
print(prev_features.shape)
print(prev_features['SK_ID_CURR'].duplicated().sum())
```

```python
df = df.merge(
    prev_features,
    on='SK_ID_CURR',
    how='left'
)
```

```python
mean_cols = [
    'PREV_AMT_APPLICATION_MEAN',
    'PREV_AMT_APPLICATION_MAX',
    'PREV_AMT_CREDIT_MEAN',
    'PREV_AMT_CREDIT_MAX',
    'PREV_AMT_ANNUITY_MEAN',
    'PREV_AMT_GOODS_PRICE_MEAN',
    'PREV_CNT_PAYMENT_MEAN',
    'PREV_AMT_DOWN_PAYMENT_MEAN',
    'PREV_RATE_DOWN_PAYMENT_MEAN'
]
```

```python
rate_cols = [
    'PREV_APPROVAL_RATE',
    'PREV_REFUSAL_RATE'
]

df[rate_cols] = df[rate_cols].fillna(0)
```

```python
# 1. Create indicator before filling nulls
df['HAS_PREV_APPLICATION'] = (
    df['PREV_APP_COUNT'].notna()
).astype(int)


# 2. No previous applications = 0 applications
df['PREV_APP_COUNT'] = df['PREV_APP_COUNT'].fillna(0)


# 3. Previous application status counts
count_cols = [
    'PREV_STATUS_APPROVED',
    'PREV_STATUS_CANCELED',
    'PREV_STATUS_REFUSED',
    'PREV_STATUS_UNUSED_OFFER'
]

df[count_cols] = df[count_cols].fillna(0)


# 4. Historical numerical features
prev_numeric_cols = [
    'PREV_AMT_APPLICATION_MEAN',
    'PREV_AMT_APPLICATION_MAX',
    'PREV_AMT_CREDIT_MEAN',
    'PREV_AMT_CREDIT_MAX',
    'PREV_AMT_ANNUITY_MEAN',
    'PREV_AMT_GOODS_PRICE_MEAN',
    'PREV_CNT_PAYMENT_MEAN',
    'PREV_AMT_DOWN_PAYMENT_MEAN',
    'PREV_RATE_DOWN_PAYMENT_MEAN'
]

df[prev_numeric_cols] = df[prev_numeric_cols].fillna(0)


# 5. Historical rates
rate_cols = [
    'PREV_APPROVAL_RATE',
    'PREV_REFUSAL_RATE'
]

df[rate_cols] = df[rate_cols].fillna(0)
```

```python
bureau_balance_fe = (
    bureau_balance_df.groupby('SK_ID_BUREAU')
    .agg(
        BUREAU_BAL_MONTHS_COUNT=('MONTHS_BALANCE', 'count'),
        BUREAU_BAL_MONTHS_MIN=('MONTHS_BALANCE', 'min'),
        BUREAU_BAL_MONTHS_MAX=('MONTHS_BALANCE', 'max')
    )
    .reset_index()
)
```

```python
print(bureau_balance_fe.head())
```

```python
status_counts = pd.crosstab(
    bureau_balance_df['SK_ID_BUREAU'],
    bureau_balance_df['STATUS']
)

# Remove the index name to avoid ambiguity
status_counts.index.name = None

# Convert SK_ID_BUREAU index into a normal column
status_counts = pd.crosstab(
    bureau_balance_df['SK_ID_BUREAU'],
    bureau_balance_df['STATUS']
).reset_index()

# Rename columns
status_counts = status_counts.rename(
    columns={
        col: f'BUREAU_BAL_STATUS_{col}_COUNT'
        for col in status_counts.columns
        if col != 'SK_ID_BUREAU'
    }
)
```

```python
print("bureau_balance_fe index name:", bureau_balance_fe.index.name)
print("bureau_balance_fe columns:", bureau_balance_fe.columns.tolist())

print("\nstatus_counts index name:", status_counts.index.name)
print("status_counts columns:", status_counts.columns.tolist())
```

```python
bureau_balance_fe = bureau_balance_fe.merge(
    status_counts,
    on='SK_ID_BUREAU',
    how='left'
)
```

```python
print(bureau_balance_fe.shape)
print(bureau_balance_fe.head())

print(
    "Duplicate SK_ID_BUREAU:",
    bureau_balance_fe['SK_ID_BUREAU'].duplicated().sum()
)
```

```python
bureau_balance_fe['BUREAU_BAL_DPD_MONTHS'] = (
    bureau_balance_fe['BUREAU_BAL_STATUS_1_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_2_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_3_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_4_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_5_COUNT']
)
```

```python
bureau_balance_fe['BUREAU_BAL_SEVERE_DPD_MONTHS'] = (
    bureau_balance_fe['BUREAU_BAL_STATUS_3_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_4_COUNT'] +
    bureau_balance_fe['BUREAU_BAL_STATUS_5_COUNT']
)
```

```python
bureau_balance_fe['BUREAU_BAL_DPD_RATE'] = (
    bureau_balance_fe['BUREAU_BAL_DPD_MONTHS'] /
    bureau_balance_fe['BUREAU_BAL_MONTHS_COUNT']
)
```

```python
bureau_enriched['AMT_CREDIT_SUM'] = (
    bureau_enriched['AMT_CREDIT_SUM']
    .fillna(bureau_enriched['AMT_CREDIT_SUM'].median())
)
```

```python
bureau_agg = bureau_enriched.groupby('SK_ID_CURR').agg(
    
    # Number of bureau credit records
    BUREAU_RECORD_COUNT=('SK_ID_BUREAU', 'count'),
    
    # Credit amounts
    BUREAU_CREDIT_SUM_MEAN=('AMT_CREDIT_SUM', 'mean'),
    BUREAU_CREDIT_SUM_MAX=('AMT_CREDIT_SUM', 'max'),
    BUREAU_CREDIT_SUM_SUM=('AMT_CREDIT_SUM', 'sum'),
    
    # Debt
    BUREAU_DEBT_MEAN=('AMT_CREDIT_SUM_DEBT', 'mean'),
    BUREAU_DEBT_MAX=('AMT_CREDIT_SUM_DEBT', 'max'),
    BUREAU_DEBT_SUM=('AMT_CREDIT_SUM_DEBT', 'sum'),
    
    # Credit limit
    BUREAU_CREDIT_LIMIT_MEAN=('AMT_CREDIT_SUM_LIMIT', 'mean'),
    BUREAU_CREDIT_LIMIT_MAX=('AMT_CREDIT_SUM_LIMIT', 'max'),
    
    # Overdue
    BUREAU_OVERDUE_MAX=('AMT_CREDIT_MAX_OVERDUE', 'max'),
    
    # Bureau balance history
    BUREAU_BAL_MONTHS_TOTAL=('BUREAU_BAL_MONTHS_COUNT', 'sum'),
    BUREAU_BAL_DPD_MONTHS_TOTAL=('BUREAU_BAL_DPD_MONTHS', 'sum'),
    BUREAU_BAL_SEVERE_DPD_MONTHS_TOTAL=(
        'BUREAU_BAL_SEVERE_DPD_MONTHS', 'sum'
    ),
    BUREAU_BAL_DPD_RATE_MEAN=('BUREAU_BAL_DPD_RATE', 'mean')
    
).reset_index()
```
